### Write the facts

```python
Hobbies = [
    ("Hakima",   "loves", "Reading"),
    ("Hakima",   "looks upto", "Malala Yousafzai"),
    ("Malala Yousafzai",   "Have", "Co-authored I Am Malala"),
    ("Jabir",   "loves", "Travelling"),
    ("Jabir",   "looks upto", "Joe Kattab"),
    ("Joe Kattab",   "Have", "Travelled 120,000km and visited around 40 countries"),
    ("Jabir ",   "loves", "Reading"),
    ("Jabir",   "looks upto", "Malala Yousafzai"),
    ("Malala Yousafzai", "Have", "Co-authored I Am Malala"),
    ("Cynthia", "loves", "Journaling"),
    ("Cynthia", "looks upto", "Anais Nin"),
    ("Anais Nin",       "Have", "kept many diaries from childhood"),
    ("Zainab", "loves", "Cooking"),
    ("Zainab", "looks upto", "Gordon Ramsay"),
    ("Gordon Ramsay", "Have", "Built an international restaurant business"),
    ("Zakir", "loves", "Football"),
    ("Zakir", "looks upto", "Christiano Ronaldo"),
    ("Christiano Ronaldo", "Have", "Won awards"),
]
```

### Turn the facts into a graph and draw it

```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph()
for subject, relation, obj in Hobbies:
    G.add_edge(subject, obj, label=relation)

pos = nx.spring_layout(G, seed=3)
plt.figure(figsize=(9, 6))
nx.draw(G, pos, with_labels=True, node_color="#cfe8ff",
        node_size=2200, font_size=10, arrows=True)
nx.draw_networkx_edge_labels(G, pos, edge_labels=nx.get_edge_attributes(G, "label"), font_size=8)
plt.show()
```
